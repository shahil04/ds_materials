## tool vs MCP
```
Not exactly. A tool and MCP are different things, but MCP can expose tools to an AI model.

Think of it this way:

Tool = WHAT the AI can do
MCP  = HOW the AI discovers and communicates with that tool

Example

Suppose you have a function:

send_email(to, subject, message)


That function is a tool.

Without MCP, your AI application could directly register that function:

AI Model
   ↓
send_email() tool
   ↓
Email API


With MCP:

AI Model
   ↓
MCP Client
   ↓
MCP Server
   ↓
send_email() tool
   ↓
Email API


The MCP server tells the AI client something like:

Available tool:
    send_email

Description:
    Send an email to a recipient.

Input:
    to: string
    subject: string
    message: string


The AI can then decide:

"I need to send an email, so I'll call send_email."

So what exactly is MCP?

MCP is a protocol — a standardized communication system.

A simplified analogy:

Tool       = a machine
API        = an interface to a service
MCP        = a standardized connection system
MCP Server = something that exposes tools/resources through MCP
MCP Client = something that connects to the MCP server


For example:

                 AI Application
                       │
                  MCP Client
                       │
                  MCP Protocol
                       │
                  MCP Server
                 /     |      \
                /      |       \
        search_files  send_email  database_query
             │            │             │
          Files        Email API      Database

The important distinction

A tool can exist without MCP.

For example:

def calculate_tax(income):
    ...


That's simply a tool/function.

MCP gives you a standardized way to make that capability available to an AI client.

So:

Tool ≠ MCP

Instead:

MCP → can expose tools

And MCP can expose more than tools too, such as resources and prompts.

If you're learning this from an AI-agent perspective, the hierarchy is roughly:

AI Model
   ↓
Agent
   ↓
Tools
   ↓
MCP (optional standardized way to connect to tools/resources)
   ↓
External systems


The biggest misconception to avoid is "MCP is a type of tool." It isn't. MCP is the protocol; tools are capabilities that can be exposed through that protocol.
```

## MCP Architecture | Model Context Protocol Architecture |


MCP - Types of Server

A local server is a program running on your own computer.

-> STDIO

A remote server is a program running on another computer (somewhere else on the network or internet) that you connect to over a network.

-> HTTP/SSE

 

Local Servers - STDIO

STDIO refers to the built-in streams every program has.

stdin (input the program reads)

stdout (output the program writes)

In MCP, these streams are used as transport layer between the client and server.
 

How does STDIO work?

The host launches the server as a subprocess on the same machine.

The host(client) writes JSON-RPC messages into the server’s STDIN

The server reads those messages, processes them, and writes back responses to it’s STDOUT
 

Benefits of the STDIO transport

Fast - data is passed directly between processes.

Secure - no open network port that could be attacked; communication is only local.

Simple → every language/runtime supports reading/writing from stdin/stdout. No extra libraries required.


---

Remote Servers - HTTP + SSE

Using HTTP allows the host to reach Servers running anywhere

Host sends JSON RPC requests using POST requests with a JSON payload

The transport supports standard HTTP auth methods(like API keys)

Remote Servers - HTTP + SSE

SSE stands for Server Sent Events, and it’s an extension of HTTP

Using SSE the server sends multiple messages to client over a single open connection

Instead of sending one large JSON blob, the server can stream chunks of data as they are ready

Ideal for long running tasks or incremental updates
