The main **LangChain components** are the building blocks used to create LLM-powered applications. They work together to process inputs, retrieve information, interact with external systems, and generate responses. ([Docs by LangChain][1])

| Component            | Purpose                                    | Example                                                                                      |
| -------------------- | ------------------------------------------ | -------------------------------------------------------------------------------------------- |
| **Models**           | Connect to LLMs or embedding models        | GPT, Claude, Gemini                                                                          |
| **Prompts**          | Create reusable prompt templates           | ChatPromptTemplate                                                                           |
| **Output Parsers**   | Convert LLM output into structured formats | JSON parser, Pydantic parser                                                                 |
| **Document Loaders** | Load data from files or websites           | PDFLoader, CSVLoader                                                                         |
| **Text Splitters**   | Break large documents into chunks          | RecursiveCharacterTextSplitter                                                               |
| **Embeddings**       | Convert text into vectors                  | OpenAI Embeddings                                                                            |
| **Vector Stores**    | Store and search embeddings                | FAISS, Chroma, Pinecone                                                                      |
| **Retrievers**       | Fetch relevant documents                   | VectorStoreRetriever                                                                         |
| **Memory**           | Maintain conversation context              | ConversationBufferMemory (legacy patterns) or message history/state in newer agent workflows |
| **Tools**            | Allow the LLM to use external resources    | Calculator, Search API, SQL Database                                                         |
| **Chains / LCEL**    | Combine multiple components into workflows | Prompt → Model → Output Parser                                                               |
| **Agents**           | Decide dynamically which tools to use      | ReAct Agent, Tool Calling Agent                                                              |

### How the components work together

```text
User Query
     │
     ▼
 Prompt Template
     │
     ▼
    LLM
     │
     ├──► Tool (Search/API)
     │
     ├──► Retriever
     │        │
     │        ▼
     │   Vector Database
     │
     ▼
Output Parser
     │
     ▼
 Final Response
```

### Example: RAG (Retrieval-Augmented Generation)

1. Load documents using a **Document Loader**.
2. Split them into chunks with a **Text Splitter**.
3. Generate **Embeddings**.
4. Store embeddings in a **Vector Store**.
5. Use a **Retriever** to find relevant chunks.
6. Send the retrieved context to the **LLM** with a **Prompt**.
7. Return the generated answer. ([Docs by LangChain][1])

The terms **AI Agent** and **Agentic AI** are related, but they are not the same.

| AI Agent                                         | Agentic AI                                                                                       |
| ------------------------------------------------ | ------------------------------------------------------------------------------------------------ |
| A single autonomous system that performs a task. | A broader approach where AI systems can plan, reason, and coordinate to achieve goals.           |
| Usually one agent with specific tools.           | Can involve one or multiple agents working together.                                             |
| Focuses on completing a defined task.            | Focuses on achieving a high-level goal through planning and adaptation.                          |
| Simpler workflows.                               | More complex, dynamic workflows with decision-making.                                            |
| Example: Customer support chatbot.               | Example: A travel planner that researches, books, and adjusts an itinerary based on constraints. |

### AI Agent

An AI agent typically:

* Receives a goal or user request.
* Chooses which tools to use (search, calculator, database, etc.).
* Performs the task.
* Returns the result.

**Example:**

**User:** "Summarize this PDF."

**Agent workflow:**

```text
User
  │
  ▼
AI Agent
  │
  ├── Read PDF
  ├── Extract text
  ├── Send to LLM
  └── Return summary
```

This is a single agent handling the entire workflow.

---

### Agentic AI

Agentic AI emphasizes **autonomous planning, reasoning, memory, and adaptation**. The system may break a goal into sub-tasks, revise its plan, and coordinate multiple specialized agents if needed.

**Example:**

**User:** "Plan a 5-day trip to Japan under ₹1,20,000."

```text
                User Goal
                    │
                    ▼
             Planner Agent
          ┌─────────┼─────────┐
          ▼         ▼         ▼
   Flight Agent Hotel Agent Budget Agent
          │         │         │
          └─────────┼─────────┘
                    ▼
            Validator Agent
                    │
             Need Changes?
             ├── Yes → Replan
             └── No  → Final Itinerary
```

The system plans, delegates work, checks results, and can revise its approach.

---

## Key differences

### AI Agent

* Solves a specific task.
* May use tools.
* Often follows a relatively straightforward workflow.
* Can be built with LangChain or LangGraph.

### Agentic AI

* Pursues broader goals.
* Plans before acting.
* Adapts when circumstances change.
* May involve multiple collaborating agents.
* Commonly implemented with frameworks such as LangGraph because they support state, branching, and iterative workflows.

### Interview-friendly explanation

> An **AI agent** is an autonomous software component that uses an LLM, tools, and memory to perform a specific task. **Agentic AI** is the broader concept of AI systems that can autonomously plan, reason, adapt, and often coordinate multiple agents to achieve complex goals. In short, an AI agent is a building block, while Agentic AI describes the overall capability or architecture that enables autonomous goal-directed behavior.


# What problem in Langchain 

https://shahil04.medium.com/langchain-vs-langgraph-agentic-ai-using-langgraph-class-3-adca66a856b4

LangGraph was created to address limitations that developers encountered when building complex, stateful AI applications with LangChain.

Here's a comparison:

| Problem in LangChain                                           | How LangGraph Solves It                                                                                 |
| -------------------------------------------------------------- | ------------------------------------------------------------------------------------------------------- |
| Linear chains are difficult to manage for complex workflows.   | Supports graph-based workflows with branching and loops.                                                |
| Limited support for long-running, multi-step agents.           | Maintains state across multiple steps and interactions.                                                 |
| Hard to implement retry, pause, or resume logic.               | Built-in support for checkpoints, retries, and resuming execution.                                      |
| Agent behavior can become unpredictable.                       | Developers explicitly define the flow between nodes, making execution easier to understand and control. |
| Managing multiple collaborating agents is cumbersome.          | Enables multi-agent systems where agents can communicate and coordinate.                                |
| Conversation and workflow state often require manual handling. | Centralized state management that is shared across graph nodes.                                         |
| Debugging complex agent workflows is challenging.              | Graph visualization and step-by-step execution improve observability.                                   |

### Example

**LangChain (Linear)**

```
User
  ↓
Retriever
  ↓
LLM
  ↓
Answer
```

This works well for straightforward RAG applications.

**LangGraph (Graph)**

```
          User
            │
            ▼
        Planner
       ↙      ↘
 Search       Calculator
      ↘      ↙
      Validator
          │
     Need Retry?
      │       │
     Yes      No
      │        │
      └──► Planner
               │
               ▼
          Final Answer
```

This type of workflow is much harder to implement cleanly with a simple chain because it involves decisions, loops, retries, and multiple tools.

### When to use each

* **Use LangChain** when you're building:

  * RAG applications
  * Chatbots
  * Simple tool calling
  * Prompt pipelines
  * Document Q&A

* **Use LangGraph** when you need:

  * AI agents
  * Multi-agent collaboration
  * Long-running workflows
  * Human approval steps
  * Complex decision trees
  * Persistent state and checkpointing

### Interview answer (short)

> LangGraph was introduced because LangChain's chain-based approach is best suited for linear workflows. As AI agents became more complex, developers needed support for branching, loops, persistent state, retries, and multi-agent coordination. LangGraph provides a graph-based execution model that enables these advanced workflows while integrating with LangChain components.
