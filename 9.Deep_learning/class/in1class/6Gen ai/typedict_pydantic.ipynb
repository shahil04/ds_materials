# TypedDict vs Pydantic (Python)

Both **TypedDict** and **Pydantic** are used to define the structure of data, but they serve different purposes.

| Feature                  | TypedDict           | Pydantic                       |
| ------------------------ | ------------------- | ------------------------------ |
| Purpose                  | Type Hinting        | Data Validation + Parsing      |
| Runtime Validation       | ❌ No                | ✅ Yes                          |
| Type Checking            | Static (mypy, IDE)  | Runtime + Static               |
| Performance              | Faster              | Slightly slower                |
| Best For                 | Simple dictionaries | APIs, LLMs, LangChain, FastAPI |
| Converts Data Types      | ❌ No                | ✅ Yes                          |
| Raises Validation Errors | ❌ No                | ✅ Yes                          |

---

# 1. TypedDict

`TypedDict` defines the expected keys and value types of a dictionary.

### Example

```python
from typing import TypedDict

class Student(TypedDict):
    name: str
    age: int

student = {
    "name": "Rahul",
    "age": 22
}

print(student["name"])
```

Output

```
Rahul
```

### Problem

```python
student = {
    "name": "Rahul",
    "age": "22"   # Wrong type
}
```

Python will **not** raise an error at runtime.

---

# 2. Pydantic

Pydantic validates data when an object is created.

### Installation

```bash
pip install pydantic
```

### Example

```python
from pydantic import BaseModel

class Student(BaseModel):
    name: str
    age: int

student = Student(
    name="Rahul",
    age=22
)

print(student)
```

Output

```
name='Rahul' age=22
```

# 1. TypedDict

`TypedDict` tells type checkers (MyPy, Pyright, VS Code) what keys a dictionary should have.

It **does not validate anything at runtime**. It only helps during development. ([typing.python.org][1])

## Example

```python
from typing import TypedDict

class Student(TypedDict):
    id: int
    name: str
    age: int

student: Student = {
    "id": 1,
    "name": "Shahil",
    "age": 22
}

print(student)
```

Output

```python
{'id': 1, 'name': 'Shahil', 'age': 22}
```

---

## Wrong Example

```python
student: Student = {
    "id": "1",    # should be int
    "name": "Shahil",
    "age": 22
}
```

Python runs this without error.

A type checker (MyPy/Pyright) reports:

```
Type of "id" should be int
```

---

## Optional Keys

```python
from typing import TypedDict, NotRequired

class Employee(TypedDict):
    id: int
    name: str
    salary: NotRequired[float]
```

Valid

```python
{
    "id": 1,
    "name": "John"
}
```

or

```python
{
    "id": 1,
    "name": "John",
    "salary": 75000
}
```

# 2. Pydantic

Pydantic creates Python classes that **validate and convert data at runtime**.

It is heavily used in **FastAPI**, APIs, configuration files, and AI/LLM projects. ([Pydantic][2])

Install

```bash
pip install pydantic
```

---

## Example

```python
from pydantic import BaseModel

class Student(BaseModel):
    id: int
    name: str
    age: int

student = Student(
    id=1,
    name="Shahil",
    age=22
)

print(student)
```

Output

```python
id=1 name='Shahil' age=22
```

---

## Automatic Type Conversion

```python
from pydantic import BaseModel

class Student(BaseModel):
    id: int
    age: int

student = Student(
    id="1",
    age="22"
)

print(student)
print(type(student.id))
```

Output

```python
id=1 age=22
<class 'int'>
```

Pydantic converts strings to integers automatically when possible.

---

## Validation Error

```python
from pydantic import BaseModel

class Student(BaseModel):
    age: int

Student(age="abc")
```

Output

```python
ValidationError:
Input should be a valid integer
```

---

## Default Values

```python
from pydantic import BaseModel

class User(BaseModel):
    name: str
    age: int = 18

user = User(name="Rahul")

print(user)
```

Output

```
name='Rahul' age=18
```

---

## Field Validation

```python
from pydantic import BaseModel, Field

class Product(BaseModel):
    price: float = Field(gt=0)

Product(price=100)
```

Invalid

```python
Product(price=-10)
```

Output

```
ValidationError
```

---

## Nested Models

```python
from pydantic import BaseModel

class Address(BaseModel):
    city: str
    country: str

class Student(BaseModel):
    name: str
    address: Address

student = Student(
    name="Shahil",
    address={
        "city": "Delhi",
        "country": "India"
    }
)

print(student)
```

# Advantages

## TypedDict

**Pros**

* Built into Python
* Very lightweight
* Great for type hints
* No external dependency
* Fast (no runtime overhead)

**Cons**

* No runtime validation
* No automatic conversion
* Cannot enforce data correctness while the program is running

---

## Pydantic

**Pros**

* Runtime validation
* Automatic type conversion
* JSON serialization/deserialization
* Nested models
* Default values
* Field constraints
* Excellent FastAPI integration

**Cons**

* Requires installing a package
* Slight validation overhead

# When to Use

### Use `TypedDict` when:

* You only need static typing.
* Working with plain dictionaries.
* Internal Python code.
* State objects in libraries like LangGraph.

Example

```python
class AgentState(TypedDict):
    question: str
    answer: str
```

---

### Use `Pydantic` when:

* Building REST APIs (FastAPI)
* Validating user input
* Parsing JSON
* Loading configuration
* LLM applications where input/output must be validated
* Production systems

Example

```python
class ChatRequest(BaseModel):
    message: str
    temperature: float
```

---

# Can They Be Used Together?

Yes. Pydantic can validate `TypedDict` definitions, combining static typing with runtime validation. ([Pydantic][2])

```python
from typing_extensions import TypedDict
from pydantic import TypeAdapter

class User(TypedDict):
    name: str
    age: int

adapter = TypeAdapter(User)

user = adapter.validate_python({
    "name": "John",
    "age": "25"
})

print(user)
```
