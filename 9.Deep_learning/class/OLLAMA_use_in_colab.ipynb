```python
import subprocess
import time
import ollama

class OllamaManager:
    def __init__(self):
        self.server_process = None

    def start_server(self):
        """Method 1: Starts the Ollama server in the background."""
        print("Starting Ollama server...")
        self.server_process = subprocess.Popen(
            ["ollama", "serve"],
            stdout=subprocess.DEVNULL,
            stderr=subprocess.DEVNULL
        )
        time.sleep(5)  # Allow time for server initialization
        print("Ollama server started.")

    def load_model(self, model_name):
        """Method 2: Pulls and loads the specified model."""
        print(f"Loading {model_name}... This might take a moment if not downloaded.")
        ollama.pull(model_name)
        print(f"{model_name} loaded successfully.")

    def ask_prompt(self, model_name, prompt):
        """Method 3: Sends a prompt to the model and returns the response."""
        response = ollama.chat(
            model=model_name,
            messages=[{'role': 'user', 'content': prompt}]
        )
        return response.message.content

# Example Usage:
# manager = OllamaManager()
# manager.start_server()
# manager.load_model('deepseek-r1:1.5b')
# reply = manager.ask_prompt('deepseek-r1:1.5b', 'Explain quantum computing in one sentence.')
# print(reply)
```

```python
# 1. Instantiate the class
manager = OllamaManager()

# 2. Start/verify the server (it's already running in the background, but we call it safely)
try:
    manager.start_server()
except Exception as e:
    print("Server is already active or encountered an issue:", e)

# 3. Load/verify the model
model = 'deepseek-r1:1.5b'
manager.load_model(model)

# 4. Ask the prompt and print the response
prompt = "hi."
response_text = manager.ask_prompt(model, prompt)

print("\n--- Model Response ---")
print(response_text)
```
