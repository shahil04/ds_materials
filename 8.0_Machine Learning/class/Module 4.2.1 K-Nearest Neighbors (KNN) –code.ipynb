## Hyper parameter Tuning

Hyperparameter tuning is the process of optimizing a machine learning model's configuration settings—such as learning rate, batch size, or model depth—before training begins, ensuring superior performance and preventing over/underfitting.

*Key Strategies for Hyperparameter Tuning:*

1. Manual Tuning: A manual process of setting hyperparameters, observing results, and refining them, usually relying on expertise.

2. Grid Search: A methodical, exhaustive search through a manually specified subset of the hyperparameter space, testing every combination.

3. Random Search: Randomly selects combinations of hyperparameters, which is often more efficient than grid search for finding optimal models, particularly when only a few hyperparameters significantly impact performance.

4. Bayesian Optimization: Uses past evaluation results to probabilistically choose the next hyperparameters to test, optimizing the search process.

Types of Cross-Validation
1. 1. Holdout Validation
In Holdout Validation the dataset is split into training and testing sets. Common splits include 70–30, 80–20 or 75–25 depending on the dataset size and problem

2. 2. LOOCV (Leave One Out Cross Validation)
In this method the model is trained on the entire dataset except for one data point which is used for testing. This process is repeated for each data point in the dataset.

3. Stratified Cross-Validation
It is a technique that ensures each fold of the cross-validation process has the same class distribution as the full dataset. This is useful for imbalanced datasets where some classes are underrepresented.

4. K-Fold Cross Validation
K-Fold Cross Validation splits the dataset into k equal-sized folds. The model is trained on k-1 folds and tested on the remaining fold. This process is repeated k times each time using a different fold for testing.

```python
from sklearn.model_selection import cross_val_score, KFold
# Defining the number of folds for cross-validation
num_folds = 5
kf = KFold(n_splits=num_folds, shuffle=True, random_state=42)

# Performing k-fold cross-validation

cross_val_results = cross_val_score(model2, x, y, cv=kf)

# Evaluation metrics
print("Cross-Validation Results (Accuracy):")
for i, result in enumerate(cross_val_results, 1):
    print(f"  Fold {i}: {result * 100:.2f}%")
    
print(f'Mean Accuracy: {cross_val_results.mean()* 100:.2f}%')

# https://www.geeksforgeeks.org/machine-learning/cross-validation-machine-learning/
```
