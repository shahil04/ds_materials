## Choosing the number of clusters

### 📊 WCSS in K-Means Clustering

## 📍 Definition

**WCSS (Within-Cluster Sum of Squares):**
It is the **sum of the squared distances** between each data point and the **centroid** of the cluster to which it belongs.

Mathematically:
[
WCSS = \sum_{k=1}^{K} \sum_{i \in C_k} ||x_i - \mu_k||^2
]

Where:

* ( K ) = Number of clusters
* ( C_k ) = Cluster ( k )
* ( x_i ) = Data point in cluster ( C_k )
* ( \mu_k ) = Centroid of cluster ( C_k )

---

## 🎯 Why WCSS is Important

* Measures **compactness** of clusters (lower WCSS = better clustering).
* Helps in **choosing the optimal number of clusters** using the **Elbow Method**.
* Indicates **how close** data points are to their cluster center.

---

## 📈 Real-World Example

| Scenario                  | Description                               | How WCSS Helps                                                                |
| ------------------------- | ----------------------------------------- | ----------------------------------------------------------------------------- |
| 🛍️ Customer Segmentation | Grouping customers by purchasing behavior | Lower WCSS means customers in each group have **similar buying patterns**     |
| 🚖 Cab Demand Prediction  | Clustering city areas by ride demand      | Lower WCSS ensures each cluster represents **geographically close** locations |
| 🏥 Patient Clustering     | Grouping patients based on symptoms       | Lower WCSS ensures each group has **similar symptom profiles**                |

---

## 📉 Elbow Method (Finding Optimal K)

To find the **optimal number of clusters**, we plot **WCSS vs. K (number of clusters)**:

* Initially, WCSS drops quickly as K increases.
* After a certain point, the decrease slows down and forms an **"elbow"**.
* The **elbow point** is considered the best ( K ).

---

✅ **Summary:**

* WCSS is a metric to measure **intra-cluster similarity**.
* It decreases as the number of clusters increases.
* Use it with the **Elbow Method** to find the best number of clusters for your K-Means model.


```python
# finding wcss value for different number of clusters
wcss = []

for i in range(1,11):
  kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
  kmeans.fit(X)

  wcss.append(kmeans.inertia_)
```

```python
# plot an elbow graph
sns.set()
plt.plot(range(1,11), wcss)
plt.title('The Elbow Point Graph')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

### 📊 Silhouette Score in K-Means (and Clustering)

The **Silhouette Score** is a powerful metric used to evaluate how **well data points are clustered**. It tells you **how similar** an object is to its **own cluster** compared to **other clusters**.

---

## 📍 Definition

The **Silhouette Score (S)** for a single data point is calculated as:

[
S = \frac{b - a}{\max(a, b)}
]

Where:

* ( a ) = Average distance of the point to all other points **in the same cluster** (intra-cluster distance)
* ( b ) = Average distance of the point to all points **in the nearest different cluster** (inter-cluster distance)

---

## 📊 Score Range & Interpretation

| Silhouette Score Range | Meaning                                                       |
| ---------------------- | ------------------------------------------------------------- |
| **+1.0**               | Perfectly clustered — far away from other clusters            |
| **0.5 – 1.0**          | Well-clustered — data points are appropriately grouped        |
| **0.0 – 0.5**          | Overlapping clusters — may not be well-separated              |
| **< 0**                | Wrong clustering — point may be assigned to the wrong cluster |

---

## 🎯 Why It’s Important

* Measures **cluster quality** based on both **cohesion** (how close points in a cluster are) and **separation** (how far clusters are apart).
* Helps validate **optimal number of clusters**.
* Works well with **K-Means, Hierarchical, DBSCAN**, etc.

---

## 📈 Real-World Examples

| Scenario                  | How Silhouette Score Helps                                        | Interpretation                          |
| ------------------------- | ----------------------------------------------------------------- | --------------------------------------- |
| 🛍️ Customer Segmentation | Evaluate if customers in each group have distinct buying patterns | **>0.6** means clusters are meaningful  |
| 🏙️ Cab Demand Clustering | Check if geographic zones are well-separated                      | **>0.5** means each zone is distinct    |
| 🩺 Patient Grouping       | Confirm patients with similar symptoms are clustered together     | **Close to 1** → clusters are very good |

---

## 🧠 Silhouette vs WCSS – Quick Comparison

| Metric               | Meaning                                               | Best Value            | Use Case                                |
| -------------------- | ----------------------------------------------------- | --------------------- | --------------------------------------- |
| **WCSS**             | Measures cluster compactness (distance to centroid)   | Lower is better       | Used for **Elbow Method**               |
| **Silhouette Score** | Measures how well-separated and cohesive clusters are | Closer to 1 is better | Used for **Cluster Quality Validation** |

---

✅ **Summary:**

* Silhouette Score is a **quality measure** for clustering.
* It combines **cohesion + separation** into one metric.
* Use it alongside **WCSS** to validate the **optimal number of clusters**.

Linkage methods in hierarchical clustering determine how the distance between two clusters is calculated before merging them. Choosing the right linkage method directly shapes the structure, density, and size of the resulting clusters.

Common Linkage Methods

• Single Linkage (Minimum): Measures the shortest distance between any single point in the first cluster and any single point in the second cluster. This method can handle non-elliptical shapes, though it often creates long, stretched-out clusters known as chaining.
• Complete Linkage (Maximum): Measures the longest distance between any single point in the first cluster and any single point in the second cluster. It produces compact, tightly bound clusters with roughly equal diameters, making it sensitive to outliers.
• Average Linkage (UPGMA): Calculates the average distance between all possible pairs of points across the two clusters. This approach balances the sensitivity of single and complete linkage, reducing the impact of outliers while avoiding extreme chaining.
• Centroid Linkage: Computes the Euclidean distance between the geometric centers (centroids) of two clusters. While intuitive, it can occasionally cause inversions in a dendrogram where clusters merge at smaller distances than previous steps.
• Ward’s Method: Merges the pair of clusters that leads to the minimum increase in total within-cluster variance. It tends to create compact, spherical clusters of similar sizes and is widely used as a default choice for numerical data.

Quick Comparison


Linkage Method	Distance Definition	Cluster Shape Tendency	Sensitivity to Outliers
Single	Minimum pairwise distance	Irregular, elongated (chaining)	High (pulls clusters together)
Complete	Maximum pairwise distance	Compact, spherical	High (distorts cluster boundaries)
Average	Mean of all pairwise distances	Moderately compact	Moderate
Centroid	Distance between cluster means	Spherical	Moderate
Ward’s	Minimum variance increase	Compact, equal-sized spheres	High


Choosing the Right Method

• Start with Ward’s or Average linkage: Use Ward’s method if you expect balanced, spherical clusters, or Average linkage for a general-purpose, outlier-resistant baseline.
• Use Single linkage for distinct shapes: Choose this when your data contains elongated paths, rings, or non-globular patterns with clear separation.
• Use Complete linkage for strict boundaries: Apply this when you need every member of a cluster to stay within a strict maximum distance from all other members.

# DBSCAN

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) is an unsupervised machine learning algorithm that groups data points based on local data density rather than pre-defined shapes or cluster counts

```python
from sklearn.cluster import DBSCAN

# Apply DBSCAN
# You may need to adjust the eps and min_samples parameters based on your data
dbscan = DBSCAN(eps=5, min_samples=5)
y_dbscan = dbscan.fit_predict(X)

print("Cluster labels for each data point:")
print(y_dbscan)
```

- **Different colored points:** Each color represents a different cluster identified by DBSCAN. These are groups of customers with similar annual income and spending scores based on the eps and min_samples parameters used.

- **Purple points (label -1):** These points represent "noise" points. DBSCAN was not able to assign these customers to any specific cluster based on the defined parameters. They are considered outliers or points in sparse areas of the data.
Density-based: Unlike K-Means and Hierarchical Clustering which form spherical or hierarchical clusters, DBSCAN finds clusters based on density. This is why the clusters might have irregular shapes and there are points not assigned to any cluster.



```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Visualize the DBSCAN clusters
plt.figure(figsize=(8, 8))
sns.scatterplot(x=X[:, 0], y=X[:, 1], hue=y_dbscan, palette='viridis', legend='full')
plt.title('DBSCAN Clusters')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.show()
```

```python
# ============================================================
# CUSTOMER SEGMENTATION - GRADIO APP
# K-Means + Hierarchical Clustering + DBSCAN
# ============================================================

import gradio as gr
import pandas as pd
import numpy as np
import pickle
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans, DBSCAN, AgglomerativeClustering
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage


# ============================================================
# 1. LOAD DATASET
# ============================================================

DATA_URL = (
    "https://media.githubusercontent.com/media/"
    "shahil04/ds_materials/refs/heads/main/"
    "8.0_Machine%20Learning/class/"
    "Module%209.1.1%20Mall_Customers.csv"
)

customer_data = pd.read_csv(DATA_URL)

# Same features used in your notebook
FEATURE_COLUMNS = [
    "Annual Income (k$)",
    "Spending Score (1-100)"
]

X = customer_data[FEATURE_COLUMNS].values


# ============================================================
# 2. LOAD TRAINED K-MEANS MODEL
# ============================================================

try:
    with open("kmeans_model.pkl", "rb") as file:
        kmeans_model = pickle.load(file)

except FileNotFoundError:
    # If model file is not available, train it automatically
    kmeans_model = KMeans(
        n_clusters=5,
        init="k-means++",
        random_state=0,
        n_init=10
    )

    kmeans_model.fit(X)


# ============================================================
# 3. CUSTOMER SEGMENT DESCRIPTION
# ============================================================

SEGMENT_DESCRIPTIONS = {
    0: "Average Income, Average Spending",
    1: "High Income, High Spending (Premium Customers)",
    2: "High Income, Low Spending",
    3: "Low Income, Low Spending",
    4: "Low Income, High Spending"
}


# ============================================================
# 4. K-MEANS CUSTOMER PREDICTION
# ============================================================

def predict_customer(income, spending):

    if income is None or spending is None:
        return "Please enter both Income and Spending Score."

    input_data = pd.DataFrame(
        [[income, spending]],
        columns=FEATURE_COLUMNS
    )

    customer_segment = int(
        kmeans_model.predict(input_data)[0]
    )

    description = SEGMENT_DESCRIPTIONS.get(
        customer_segment,
        "Unknown Customer Segment"
    )

    return (
        f"### Predicted Customer Segment\n\n"
        f"**Cluster:** {customer_segment}\n\n"
        f"**Segment:** {description}\n\n"
        f"**Annual Income:** {income} k$\n\n"
        f"**Spending Score:** {spending}"
    )


# ============================================================
# 5. K-MEANS VISUALIZATION
# ============================================================

def kmeans_visualization(n_clusters):

    model = KMeans(
        n_clusters=int(n_clusters),
        init="k-means++",
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)
    centers = model.cluster_centers_

    fig, ax = plt.subplots(figsize=(9, 6))

    scatter = ax.scatter(
        X[:, 0],
        X[:, 1],
        c=labels,
        s=50
    )

    ax.scatter(
        centers[:, 0],
        centers[:, 1],
        s=200,
        marker="X",
        label="Centroids"
    )

    ax.set_title(
        f"K-Means Customer Segmentation - {n_clusters} Clusters"
    )

    ax.set_xlabel("Annual Income (k$)")
    ax.set_ylabel("Spending Score (1-100)")

    ax.legend()
    ax.grid(alpha=0.2)

    return fig


# ============================================================
# 6. ELBOW METHOD
# ============================================================

def elbow_graph():

    wcss = []

    for i in range(1, 11):

        model = KMeans(
            n_clusters=i,
            init="k-means++",
            random_state=42,
            n_init=10
        )

        model.fit(X)

        wcss.append(model.inertia_)

    fig, ax = plt.subplots(figsize=(9, 6))

    ax.plot(
        range(1, 11),
        wcss,
        marker="o"
    )

    ax.set_title("Elbow Method - Finding Optimal K")

    ax.set_xlabel("Number of Clusters")

    ax.set_ylabel("WCSS / Inertia")

    ax.grid(alpha=0.2)

    return fig


# ============================================================
# 7. HIERARCHICAL CLUSTERING
# ============================================================

def hierarchical_visualization(n_clusters):

    model = AgglomerativeClustering(
        n_clusters=int(n_clusters),
        metric="euclidean",
        linkage="ward"
    )

    labels = model.fit_predict(X)

    fig, ax = plt.subplots(figsize=(9, 6))

    ax.scatter(
        X[:, 0],
        X[:, 1],
        c=labels,
        s=50
    )

    ax.set_title(
        f"Hierarchical / Agglomerative Clustering - "
        f"{n_clusters} Clusters"
    )

    ax.set_xlabel("Annual Income (k$)")
    ax.set_ylabel("Spending Score (1-100)")

    ax.grid(alpha=0.2)

    return fig


# ============================================================
# 8. DENDROGRAM
# ============================================================

def hierarchical_dendrogram():

    fig, ax = plt.subplots(figsize=(12, 7))

    linked = linkage(
        X,
        method="ward"
    )

    dendrogram(
        linked,
        ax=ax,
        truncate_mode="lastp",
        p=30,
        leaf_rotation=90,
        leaf_font_size=8
    )

    ax.set_title(
        "Hierarchical Clustering - Dendrogram"
    )

    ax.set_xlabel("Customer Groups")

    ax.set_ylabel("Euclidean Distance")

    ax.grid(alpha=0.2)

    return fig


# ============================================================
# 9. DBSCAN
# ============================================================

def dbscan_visualization(eps, min_samples):

    model = DBSCAN(
        eps=float(eps),
        min_samples=int(min_samples)
    )

    labels = model.fit_predict(X)

    fig, ax = plt.subplots(figsize=(9, 6))

    ax.scatter(
        X[:, 0],
        X[:, 1],
        c=labels,
        s=50
    )

    ax.set_title(
        f"DBSCAN Clustering "
        f"(eps={eps}, min_samples={min_samples})"
    )

    ax.set_xlabel("Annual Income (k$)")
    ax.set_ylabel("Spending Score (1-100)")

    ax.grid(alpha=0.2)

    return fig


# ============================================================
# 10. MODEL COMPARISON TABLE
# ============================================================

def model_comparison():

    results = []

    # -----------------------------
    # K-MEANS
    # -----------------------------

    km = KMeans(
        n_clusters=5,
        init="k-means++",
        random_state=42,
        n_init=10
    )

    km_labels = km.fit_predict(X)

    km_silhouette = silhouette_score(
        X,
        km_labels
    )

    results.append({
        "Algorithm": "K-Means",
        "Parameters": "n_clusters=5",
        "Clusters": len(set(km_labels)),
        "Noise Points": 0,
        "Silhouette Score": round(km_silhouette, 4),
        "Main Idea": "Centroid-based"
    })


    # -----------------------------
    # HIERARCHICAL
    # -----------------------------

    hc = AgglomerativeClustering(
        n_clusters=5,
        metric="euclidean",
        linkage="ward"
    )

    hc_labels = hc.fit_predict(X)

    hc_silhouette = silhouette_score(
        X,
        hc_labels
    )

    results.append({
        "Algorithm": "Hierarchical",
        "Parameters": "n_clusters=5, linkage=ward",
        "Clusters": len(set(hc_labels)),
        "Noise Points": 0,
        "Silhouette Score": round(hc_silhouette, 4),
        "Main Idea": "Tree / Dendrogram"
    })


    # -----------------------------
    # DBSCAN
    # -----------------------------

    db = DBSCAN(
        eps=5,
        min_samples=5
    )

    db_labels = db.fit_predict(X)

    unique_labels = set(db_labels)

    number_of_clusters = len(
        unique_labels - {-1}
    )

    noise_points = list(db_labels).count(-1)

    # Silhouette requires at least 2 clusters
    if number_of_clusters >= 2:

        mask = db_labels != -1

        db_silhouette = silhouette_score(
            X[mask],
            db_labels[mask]
        )

    else:
        db_silhouette = np.nan

    results.append({
        "Algorithm": "DBSCAN",
        "Parameters": "eps=5, min_samples=5",
        "Clusters": number_of_clusters,
        "Noise Points": noise_points,
        "Silhouette Score": (
            round(db_silhouette, 4)
            if not np.isnan(db_silhouette)
            else "N/A"
        ),
        "Main Idea": "Density-based"
    })

    return pd.DataFrame(results)


# ============================================================
# 11. CLUSTER STATISTICS
# ============================================================

def cluster_statistics():

    model = KMeans(
        n_clusters=5,
        init="k-means++",
        random_state=42,
        n_init=10
    )

    labels = model.fit_predict(X)

    data = customer_data.copy()

    data["KMeans_Cluster"] = labels

    result = (
        data
        .groupby("KMeans_Cluster")[
            FEATURE_COLUMNS
        ]
        .agg(
            ["mean", "median", "count"]
        )
        .round(2)
    )

    return result


# ============================================================
# 12. TRAINING DATA WITH LABELS
# ============================================================

def training_table():

    km = KMeans(
        n_clusters=5,
        init="k-means++",
        random_state=42,
        n_init=10
    )

    km_labels = km.fit_predict(X)

    hc = AgglomerativeClustering(
        n_clusters=5,
        metric="euclidean",
        linkage="ward"
    )

    hc_labels = hc.fit_predict(X)

    db = DBSCAN(
        eps=5,
        min_samples=5
    )

    db_labels = db.fit_predict(X)

    result = customer_data[
        ["CustomerID",
         "Gender",
         "Age",
         "Annual Income (k$)",
         "Spending Score (1-100)"]
    ].copy()

    result["KMeans"] = km_labels

    result["Hierarchical"] = hc_labels

    result["DBSCAN"] = db_labels

    return result


# ============================================================
# 13. GRADIO CSS
# ============================================================

css = """
.gradio-container {
    max-width: 1200px !important;
    margin: auto;
}

h1 {
    text-align: center;
}
"""


# ============================================================
# 14. GRADIO APPLICATION
# ============================================================

with gr.Blocks(
    title="Customer Segmentation Analysis",
    css=css,
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown(
        """
        # 🛍️ Customer Segmentation Analysis

        ### Machine Learning Clustering Demonstration

        This application demonstrates:

        **K-Means | Hierarchical Clustering | DBSCAN**

        Dataset:
        **Mall Customers**

        Features:

        - Annual Income (k$)
        - Spending Score (1-100)
        """
    )


    # ========================================================
    # TAB 1 - PREDICTION
    # ========================================================

    with gr.Tab("🎯 Customer Prediction"):

        gr.Markdown(
            """
            ## Predict Customer Segment

            Enter the customer's:

            - Annual Income
            - Spending Score

            The trained **K-Means model** will predict the cluster.
            """
        )

        with gr.Row():

            income_input = gr.Number(
                label="Annual Income (k$)",
                value=50,
                minimum=0
            )

            spending_input = gr.Number(
                label="Spending Score (1-100)",
                value=50,
                minimum=0,
                maximum=100
            )

        predict_button = gr.Button(
            "🔮 Customer Segment Predict",
            variant="primary"
        )

        prediction_output = gr.Markdown()

        predict_button.click(
            fn=predict_customer,
            inputs=[
                income_input,
                spending_input
            ],
            outputs=prediction_output
        )


    # ========================================================
    # TAB 2 - K-MEANS
    # ========================================================

    with gr.Tab("🔵 K-Means"):

        gr.Markdown(
            """
            ## K-Means Clustering

            K-Means divides customers into **K clusters**
            using distance from cluster centroids.
            """
        )

        kmeans_clusters = gr.Slider(
            minimum=2,
            maximum=10,
            value=5,
            step=1,
            label="Number of Clusters (K)"
        )

        kmeans_plot = gr.Plot(
            label="K-Means Visualization"
        )

        kmeans_button = gr.Button(
            "Generate K-Means Clusters"
        )

        kmeans_button.click(
            fn=kmeans_visualization,
            inputs=kmeans_clusters,
            outputs=kmeans_plot
        )

        gr.Markdown(
            """
            ### 💡 Student Concept

            **Centroid:** Center point of a cluster.

            **WCSS:** Within Cluster Sum of Squares.

            Lower WCSS generally means points are closer
            to their cluster centroids.
            """
        )


    # ========================================================
    # TAB 3 - ELBOW
    # ========================================================

    with gr.Tab("📈 Elbow Method"):

        gr.Markdown(
            """
            ## Elbow Method

            The Elbow Method helps us select an appropriate
            number of clusters for K-Means.
            """
        )

        elbow_plot = gr.Plot()

        elbow_button = gr.Button(
            "Generate Elbow Graph",
            variant="primary"
        )

        elbow_button.click(
            fn=elbow_graph,
            outputs=elbow_plot
        )


    # ========================================================
    # TAB 4 - HIERARCHICAL
    # ========================================================

    with gr.Tab("🌳 Hierarchical Clustering"):

        gr.Markdown(
            """
            ## Hierarchical / Agglomerative Clustering

            Hierarchical clustering builds a hierarchy of
            clusters.

            The **dendrogram** helps visualize how clusters
            are merged.
            """
        )

        hierarchical_clusters = gr.Slider(
            minimum=2,
            maximum=10,
            value=5,
            step=1,
            label="Number of Clusters"
        )

        hierarchical_plot = gr.Plot()

        hierarchical_button = gr.Button(
            "Generate Hierarchical Clusters"
        )

        hierarchical_button.click(
            fn=hierarchical_visualization,
            inputs=hierarchical_clusters,
            outputs=hierarchical_plot
        )

        gr.Markdown("### 🌳 Dendrogram")

        dendrogram_plot = gr.Plot()

        dendrogram_button = gr.Button(
            "Generate Dendrogram"
        )

        dendrogram_button.click(
            fn=hierarchical_dendrogram,
            outputs=dendrogram_plot
        )


    # ========================================================
    # TAB 5 - DBSCAN
    # ========================================================

    with gr.Tab("🟣 DBSCAN"):

        gr.Markdown(
            """
            ## DBSCAN

            DBSCAN is a **density-based clustering algorithm**.

            It can identify:

            - Dense regions
            - Clusters
            - Noise / Outliers
            """
        )

        with gr.Row():

            eps_input = gr.Number(
                label="eps",
                value=5
            )

            min_samples_input = gr.Number(
                label="min_samples",
                value=5,
                minimum=1
            )

        dbscan_plot = gr.Plot()

        dbscan_button = gr.Button(
            "Generate DBSCAN Clusters"
        )

        dbscan_button.click(
            fn=dbscan_visualization,
            inputs=[
                eps_input,
                min_samples_input
            ],
            outputs=dbscan_plot
        )


    # ========================================================
    # TAB 6 - MODEL COMPARISON
    # ========================================================

    with gr.Tab("📊 Algorithm Comparison"):

        gr.Markdown(
            """
            # Clustering Algorithm Comparison

            This table allows students to compare the
            three clustering algorithms used in this lesson.
            """
        )

        comparison_table = gr.Dataframe(
            headers=[
                "Algorithm",
                "Parameters",
                "Clusters",
                "Noise Points",
                "Silhouette Score",
                "Main Idea"
            ],
            interactive=False,
            wrap=True
        )

        comparison_button = gr.Button(
            "🚀 Train & Compare All Models",
            variant="primary"
        )

        comparison_button.click(
            fn=model_comparison,
            outputs=comparison_table
        )


    # ========================================================
    # TAB 7 - TRAINING DATA
    # ========================================================

    with gr.Tab("📚 Training Dataset"):

        gr.Markdown(
            """
            # Training Dataset + Cluster Labels

            This table shows how the **same customer data**
            can receive different cluster labels from:

            - K-Means
            - Hierarchical Clustering
            - DBSCAN

            ⚠️ Cluster numbers are algorithm-specific.
            Cluster `0` in K-Means does not necessarily mean
            the same group as Cluster `0` in DBSCAN.
            """
        )

        training_output = gr.Dataframe(
            interactive=False,
            wrap=True
        )

        training_button = gr.Button(
            "Show Training Data + All Cluster Labels",
            variant="primary"
        )

        training_button.click(
            fn=training_table,
            outputs=training_output
        )


    # ========================================================
    # TAB 8 - CLUSTER STATISTICS
    # ========================================================

    with gr.Tab("📋 Cluster Statistics"):

        gr.Markdown(
            """
            # K-Means Cluster Characteristics

            Mean, median and number of customers
            in each cluster.
            """
        )

        statistics_output = gr.Dataframe(
            interactive=False
        )

        statistics_button = gr.Button(
            "Analyze Customer Clusters",
            variant="primary"
        )

        statistics_button.click(
            fn=cluster_statistics,
            outputs=statistics_output
        )


    # ========================================================
    # FOOTER
    # ========================================================

    gr.Markdown(
        """
        ---

        ### 👨‍🏫 Machine Learning Teaching Demo

        **Topic:** Customer Segmentation using Clustering

        **Algorithms:**

        `K-Means` • `Hierarchical Clustering` • `DBSCAN`

        **Features:**

        `Annual Income` • `Spending Score`
        """
    )


# ============================================================
# 15. LAUNCH
# ============================================================
if __name__ == "__main__":
    demo.launch()
```
