For students who are new to DSA, I recommend teaching in this order:

| ---- | -------------------------- | ------------------------------ |
| 1    | Number problems + patterns | Logical thinking               |
| 2    | Strings                    | Character manipulation         |

## **Important concepts to introduce alongside DSA**

After students understand these basic problems, introduce:

- Time Complexity: \\(O(1), O(n), O(n^2), O(\log n)\\)
- Space Complexity
- Big O Notation
- Array vs Linked List
- Stack and Queue
- Hashing
- Two Pointer Technique

**My recommendation for classroom practice**

Don't teach all 100 questions with ready-made solutions initially.

1. Explain the problem using a real-world example.
2. Ask students to identify the input and expected output.
3. Let them write pseudocode.
4. Ask them to solve without built-in shortcuts.
5. Discuss time and space complexity.
6. Give a similar problem as homework.

This approach develops actual problem-solving skills instead of memorizing Python code.
